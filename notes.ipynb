```python

courses = ['Machine Learning', 'SQL Basics', 'Python Pro', ' Data Analytics', 'Cloud Databases']
num_of_students = [50, 90, 30, 130, 16]

plt.figure(figsize=(10, 8))
plt.bar(courses, num_of_students, color='purple', width=0.5)
plt.title('Number of Student Per Class')
plt.xlabel('Class Name')
plt.ylabel('Number of Students')

plt.show()
```

```python
plt.figure(figsize=(10, 8))
plt.barh(courses, num_of_students, color='purple', height=0.5)
plt.title('Number of Student Per Class')
plt.ylabel('Class Name')
plt.xlabel('Number of Students')
```

```python
courses = ['Machine Learning', 'SQL Basics', 'Python Pro', ' Data Analytics', 'Cloud Databases']
course_group1 = [50, 90, 30, 130, 16]
course_group2 = [25, 60, 180, 12, 50]

plt.figure(figsize=(10, 8))
plt.bar(courses, course_group1, width=0.3, label='Group 1')

plt.bar(courses, course_group2, width=0.3, bottom=course_group1, label='Group 2')


plt.title('Number of Student Per Class')
plt.xlabel('Class Name')
plt.ylabel('Number of Students')
plt.legend()

plt.show()
```

```python
years = ['2016', '2017', '2018', '2019']
rice_price = [57, 50, 47, 30]
wheat_price = [43, 50, 53, 70]

plt.figure(figsize=(10, 8))

plt.bar(years, rice_price, label='Rice')
plt.bar(years, wheat_price, bottom=rice_price, label='Wheat')

plt.xlabel('Years', fontsize=12)
plt.ylabel('Preference (%)', fontsize=12)
plt.title('The Preference of Rice and Wheat', fontsize=20)
plt.legend()
plt.show()
```

```python
years = ['2016', '2017', '2018', '2019']
rice =  np.array([57, 50, 47, 30])
wheat =  np.array([43, 50, 53, 70])
mustard =  np.array([10, 16, 13, 15])

plt.figure(figsize=(10, 8))

plt.bar(years, rice, label='Rice', color='purple', width=0.3)
plt.bar(years, wheat, bottom=rice, label='Wheat', width=0.3)
plt.bar(years, mustard, bottom=rice+wheat, label='Mustard seeds', color='pink', width=0.3)

plt.xlabel('Years')
plt.ylabel('Preference (%)',)
plt.title('The results of Rice,Wheat & Mustard seeds', fontsize=20)
plt.legend()
plt.show()
```

```python
courses = ['Machine Learning', 'SQL Basics', 'Python Pro', ' Data Analytics', 'Cloud Databases']
course_group1 = [50, 90, 30, 130, 16]
course_group2 = [25, 60, 180, 12, 50]
width = 0.3

# calcuaate colmn placment. creating a list/array with the index values for each category
x = np.arange(len(courses))
plt.bar(x - width/2, course_group1, width=width, label='Group1', color='purple')
plt.bar(x + width/2, course_group2, width=width, label='Group2', color='pink')
plt.xticks(x, courses, rotation=45)
plt.legend()
plt.show()
```
